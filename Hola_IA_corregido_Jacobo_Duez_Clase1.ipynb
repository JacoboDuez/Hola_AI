## Paso 1 · Percepción

El agente necesita guardar lo que sabe del mundo. Cada dato que guarda con un nombre es una **variable**.

Estas tres líneas son la percepción del agente: son todo lo que sabe del exterior.


```python
# PERCEPCIÓN
probabilidad_lluvia = 70
voy_a_salir = True
tengo_paraguas = True

print("Probabilidad de lluvia:", probabilidad_lluvia)
print("¿Voy a salir?:", voy_a_salir)
print("¿Tengo paraguas?:", tengo_paraguas)
```

**Prueba esto.** Cambia el `70` por `20` y vuelve a ejecutar la celda.

El agente todavía no decide nada; sólo repite lo que sabe.

## Paso 2 · Decisión

El rombo del diagrama de flujo se escribe con `if`.

Lo que va debajo, con sangría, sólo ocurre cuando la respuesta a la pregunta es sí.

El `50` es un **umbral**. No vino de ningún lado: lo escogimos nosotros, y esa elección ya es parte del diseño del agente.


```python
# DECISIÓN
if probabilidad_lluvia > 50:
    decision = "Llevar paraguas"
else:
    decision = "No llevar paraguas"
```

## Paso 3 · Acción

La acción del agente, por ahora, es decir lo que decidió.


```python
# ACCIÓN
print(decision)
```

## El agente completo

Cada bloque tiene nombre, y el nombre ya lo conoces.


```python
# PERCEPCIÓN
probabilidad_lluvia = 70

# DECISIÓN
if probabilidad_lluvia > 50:
    decision = "Llevar paraguas"
else:
    decision = "No llevar paraguas"

# ACCIÓN
print(decision)
```

## Paso 4 · El agente tiene un hueco

Ejecuta la celda de abajo y lee con cuidado lo que responde.

El agente dice que lleves un paraguas que no tienes.

**El hueco no está en el código:** está en cómo planteamos el problema.


```python
# se obtiene las percepcion a traves de la inicializacion de las variables
#
probabilidad_lluvia = 90
tengo_paraguas = False

if probabilidad_lluvia > 50:
    decision = "Llevar paraguas"
else:
    decision = "No llevar paraguas"

print(decision)
```

### Te toca

Completa la celda de abajo para que el agente considere si de verdad tienes paraguas.

Cuando la probabilidad sea alta y **no** tengas paraguas, la decisión debe ser `"Buscar alternativa"`.

**Pista:** puedes poner un `if` dentro de otro `if`, con más sangría.


```python
probabilidad_lluvia = 90
tengo_paraguas = False

if probabilidad_lluvia > 50:
    # TODO: aquí falta preguntar si tengo paraguas
    if not tengo_paraguas:
      decision = "Llevar paraguas"
else:
    decision = "No llevar paraguas"

print(decision)
```

### Solución de referencia

Ejecuta esta celda después de intentar resolver el ejercicio.


```python
probabilidad_lluvia = 90
tengo_paraguas = False

if probabilidad_lluvia > 50:
    if tengo_paraguas:
        decision = "Llevar paraguas"
    else:
        decision = "Buscar alternativa"
else:
    decision = "No llevar paraguas"

print(decision)
```

## Paso 5 · Repetición

En el diagrama de flujo vimos tres estructuras: secuencia, condición y **repetición**.

Hasta ahora el agente decide una sola vez. Un agente real revisaría el pronóstico varias veces antes de salir.

Eso es un ciclo: un tramo que se repite.


```python
pronosticos = [20, 35, 60, 85]  # pronóstico de cada hora antes de salir

for probabilidad_lluvia in pronosticos:
    if probabilidad_lluvia > 50:
        decision = "Llevar paraguas"
    else:
        decision = "No llevar paraguas"

    print("Probabilidad", probabilidad_lluvia, "→", decision)
```

**Fíjate en algo.** El agente cambia de opinión a media lista.

Un agente reactivo decide con el dato del momento y nada más; uno con memoria recordaría que ya venía subiendo.
